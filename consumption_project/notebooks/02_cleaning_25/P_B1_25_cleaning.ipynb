```python
# Doing for learning purposes as alternative to the non-zero counts and percentage table. 
# Frequency of rows where cement_42_5r is non-zero.
frequency = (P_B1_25_clean["cement_42_5r"] != 0).sum()
print(f"Frequency of rows where 'cement_42_5r' is non-zero: {frequency}")
```

```python
# Show the row where cement_42_5r is non-zero.
non_zero_rows = P_B1_25_clean[P_B1_25_clean["cement_42_5r"] != 0]
print(non_zero_rows)
```

```python
# Merging cement_42_5r and cement_42_5r_1 and drop the cement_42_5r_1 column.
P_B1_25_clean["cement_42_5r"] = P_B1_25_clean["cement_42_5r"] + P_B1_25_clean["cement_42_5r_1"]
P_B1_25_clean.drop(columns=["cement_42_5r_1"], inplace=True)
```

```python
P_B1_25_clean.columns.tolist()
```

```python
# Frequency of rows where the condition is that product cement_42_5r and cement_3a_42 are both !=0 in the same row.
condition = (P_B1_25_clean["cement_42_5r"] != 0) & (P_B1_25_clean["cement_3a_42"] != 0)
frequency = condition.sum()
print(f"Frequency of rows where both cement_42_5r and cement_3a_42 are != 0: {frequency}")
```

```python
# We see that there are 0 rows where both cement_42_5r and cement_a_42 are != 0. 
# Which means that there are no loads where both of these products are present in the same row.
# So, all product with cement use cement_42_5r or cement_a_42.
```

```python
# How many non-zero values each column has and what percentage of all rows that represents.
non_zero_counts = (P_B1_25_clean != 0.0).sum()
non_zero_pct = non_zero_counts / len(P_B1_25_clean) * 100

pd.DataFrame({
    "non_zero_count": non_zero_counts,
    "non_zero_pct": non_zero_pct.round(2)
})
```

```python
# we add a new column where all values will be P_B1 because in the future merging we need to identify from which plant the load is come from. 
# Inserting a new column and value. 
# P_B1_25_clean["plant"] = "P_B1"
# Inserting a column position too. 
P_B1_25_clean.insert(1, "plant", "P_B1")
```

```python
# The final datsets (all plants and both years merged) will have the following columns in this order:
# ['date', 'plant', 'product', 'volume', 'cement_42_5r', 'cement_3a_42', 'cement_4av', 'arid_04', 'arid_8_14', 'arid_12_20', 'water', 'ad_complast_mr', 'ad_dynamon_sx', 'ad_aut_dynamon', 'ad_auramix', 'corrective_02']
# This step could be done in the merge file, but the column name need to be the same in all plants and years, so we do it here. So instead of doing just this step we order them too.
final_order = [
    'date',
    'plant',
    'product',
    'volume',
    'cement_42_5r',
    'cement_3a_42',
    'arid_04',
    'arid_8_14',
    'arid_12_20',
    'water',
    'ad_dynamon_sx',
    'ad_auramix',
    'corrective_02'
]

P_B1_25_clean = P_B1_25_clean[final_order]
print(P_B1_25_clean.columns.tolist())
```
